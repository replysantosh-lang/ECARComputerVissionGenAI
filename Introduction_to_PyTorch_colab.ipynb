# Introduction to PyTorch

PyTorch is an open-source machine learning framework that is widely used for deep learning applications. In this section, we will learn the fundamentals of PyTorch.

### 1. Tensors and Gradients

#### 1.1 Tensors

Tensors are the fundamental data structures in PyTorch, similar to NumPy arrays but with the added capability of utilizing GPUs for faster computations.

Here we create a simple tensor in PyTorch.

```python
import torch

# Create a tensor
x = torch.tensor([1.0, 2.0, 3.0])
print(x)
```

#### 1.2 Gradients and Autograd

The autograd package in PyTorch automatically computes the gradients for tensors involved in a computation. This is particularly useful for implementing deep learning models, where backpropagation is necessary.

**Basic Autograd Example**

In this example, we create tensors, build a computational graph, and compute the gradients automatically.

```python
# Create tensors with requires_grad=True to track computations
x = torch.tensor(1., requires_grad=True)
w = torch.tensor(2., requires_grad=True)
b = torch.tensor(3., requires_grad=True)

# Build a computational graph (y = w * x + b)
y = w * x + b

# Compute gradients (dy/dx, dy/dw, dy/db)
y.backward()

# Print out the gradients
print('Gradient of x:', x.grad)
print('Gradient of w:', w.grad)
print('Gradient of b:', b.grad)
```

**Explanation:**
- `requires_grad=True`: Tells PyTorch to track all operations on the tensor.
- `.backward()`: Computes the gradient.
- `.grad`: Accesses the computed gradients.

**Gradient Computation in a Neural Network Layer**

In this example, we use PyTorch's nn.Linear layer and demonstrate how gradients are computed for more complex operations.

```python
import torch.nn as nn

# Create random input and output tensors
x = torch.randn(10, 3)
y = torch.randn(10, 2)

# Define a linear layer
linear = nn.Linear(3, 2)

# Forward pass
pred = linear(x)

# Define a loss function and compute the loss
criterion = nn.MSELoss()
loss = criterion(pred, y)

print('Initial loss:', loss.item())

# Backward pass to compute gradients
loss.backward()

# Print gradients of the weights and biases
print('Gradient of weights:', linear.weight.grad)
print('Gradient of biases:', linear.bias.grad)

# Perform one step of gradient descent
optimizer = torch.optim.SGD(linear.parameters(), lr=0.01)
optimizer.step()

# Check loss after one step optimization
pred = linear(x)
loss = criterion(pred, y)
print('Loss after one step optimization:', loss.item())
```

```python
print(f'Actual y is {y}')
print(f'Predicted y is {pred}')
```

**Explanation:**
- Linear Layer: A fully connected layer (`nn.Linear`).
- Loss Function: Measures the difference between the predicted and true output.
- Backpropagation: The `.backward()` function calculates the gradients of the loss with respect to each parameter.
- Optimizer: Updates the parameters based on the gradients using the `.step()` function.

#### 1.3 Loading Data from NumPy

PyTorch seamlessly integrates with NumPy. You can easily convert between PyTorch tensors and NumPy arrays.

```python
import numpy as np

# Create a NumPy array
np_array = np.array([[1, 2], [3, 4]])

# Convert NumPy array to PyTorch tensor
torch_tensor = torch.from_numpy(np_array)
print('Tensor:', torch_tensor)

# Convert back to NumPy
back_to_np = torch_tensor.numpy()
print('Numpy Array:', back_to_np)
```

**Explanation:**
- `torch.from_numpy()`: Converts a NumPy array to a PyTorch tensor.
- `.numpy()`: Converts a PyTorch tensor back to a NumPy array.

### 2. Getting started with Deep Learning

#### 2.1 Input Pipeline

In deep learning, managing and processing datasets efficiently is critical. PyTorch provides the `DataLoader` class to load data in mini-batches, shuffle the data, and use multiprocessing for speed.

In this example, we use PyTorch's built-in dataset loader to download and preprocess the CIFAR-10 dataset.

```python
import torchvision
import torchvision.transforms as transforms

# Download and construct CIFAR-10 training dataset
train_dataset = torchvision.datasets.CIFAR10(root='./data',
                                             train=True,
                                             transform=transforms.ToTensor(),
                                             download=True)

# Create a data loader for batching
train_loader = torch.utils.data.DataLoader(dataset=train_dataset,
                                           batch_size=64,
                                           shuffle=True)

# Iterate through the data loader
data_iter = iter(train_loader)
images, labels = next(data_iter)
print(f'Image batch dimensions: {images.size()}')
print(f'Label batch dimensions: {labels.size()}')
```

```python
train_dataset.data.shape
```

**Explanation:**
- `DataLoader`: Manages the loading of data in batches.
- `transforms.ToTensor()`: Converts images to PyTorch tensors for easier processing.
- `shuffle=True`: Shuffles the dataset to reduce bias during training.

#### 2.2 Pre-trained Model

PyTorch provides a large library of pre-trained models through `torchvision.models`. These models can be fine-tuned for your specific tasks, which significantly speeds up the training process.

We load a pre-trained ResNet-18 model and replace the final layer to adapt it for a custom classification task.

```python
import torchvision.models as models
import torch.nn as nn

# Load the pretrained ResNet-18 model
resnet = models.resnet18(pretrained=True)

# Freeze all layers except the final layer
for param in resnet.parameters():
    param.requires_grad = False

# Replace the final fully connected layer to match the number of output classes (e.g., 100 classes)
resnet.fc = nn.Linear(resnet.fc.in_features, 100)

# Forward pass example
images = torch.randn(64, 3, 224, 224)
outputs = resnet(images)
print(f'Output size: {outputs.size()}')  # Should be (64, 100)
```

**Explanation:**
- Pre-trained Model: Loads a pre-trained model from PyTorch’s model zoo.
- Freezing Layers: Disables training on earlier layers while fine-tuning only the final fully connected layer.
- Customizing Output: Modifies the model to fit the specific number of classes in your dataset.

#### 2.3 Saving and Loading Models

```python
# Save the entire model
torch.save(resnet, 'resnet_model.pth')

# Save only the model parameters (recommended)
torch.save(resnet.state_dict(), 'resnet_params.pth')
```

```python
# Load the entire model
loaded_model = torch.load('resnet_model.pth')

# Load the model parameters into an identical architecture
resnet = models.resnet18()
resnet.fc = nn.Linear(resnet.fc.in_features, 100)  # Make sure architecture matches
resnet.load_state_dict(torch.load('resnet_params.pth'))
```

**Explanation:**
- `torch.save()`: Saves the model or its parameters to disk.
- `torch.load()`: Loads the entire model or the saved parameters back into memory.
- Best Practice: It is recommended to save and load model parameters using state_dict() as it is more flexible.
